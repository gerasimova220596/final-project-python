Сразу переведем столбец "Дата" в правильный формат

```python
# Преобразуем текстовые даты в формат datetime
df['Дата'] = pd.to_datetime(df['Дата'], format='%Y-%m-%d')

# Проверяем результат преобразования
df['Дата'].head()
```

Сгруппируйте данные по дате, посчитайте количество продаж

```python
# Суммируем количество проданных единиц за каждую дату
grouped_df = (
    df.groupby('Дата')['Количество']
    .sum()
    .rename('Количество продаж')
    .to_frame()
    .sort_index()
)
```

Вывести несколько первых строк сгруппированных данных

```python
# Проверяем первые пять дней после группировки
grouped_df.head()
```

Нарисуйте график продаж у `grouped_df`

```python
# Строим график ежедневного количества продаж
ax = grouped_df.plot(
    figsize=(14, 5),
    color='steelblue',
    linewidth=1.5,
    legend=False
)

ax.set_title('Динамика ежедневных продаж')
ax.set_xlabel('Дата')
ax.set_ylabel('Количество проданных единиц')
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

Найдите строку, у которой максимальный выброс по количеству продаж (нужно найти выброс у `df`)

```python
# Находим максимальное количество в одной исходной записи
max_quantity = df['Количество'].max()

# Показываем все строки с этим максимумом
max_rows = df.loc[df['Количество'].eq(max_quantity)]

print('Максимальное количество:', max_quantity)
max_rows
```

Найдите топовый товар по продажам по средам за июнь, июль, август у 3 склада

```python
# Отбираем склад 3, летние месяцы 2018 года и среды
mask = (
    (df['Склад'] == 3)
    & (df['Дата'].dt.year == 2018)
    & (df['Дата'].dt.month.isin([6, 7, 8]))
    & (df['Дата'].dt.dayofweek == 2)
)

# Суммируем проданные единицы по каждому товару
top_products = (
    df.loc[mask]
    .groupby('Номенклатура')['Количество']
    .sum()
    .sort_values(ascending=False)
)

# Выводим лидеров и все товары, разделившие первое место
print('Топ-5 товаров:')
display(top_products.head())

print('Товар с максимальными продажами:')
display(top_products[top_products.eq(top_products.max())])
```

А также отдельно график температуры.

```python
# Определяем период для загрузки погоды
print('Начало:', df['Дата'].min().date())
print('Конец:', df['Дата'].max().date())
```

```python
from pathlib import Path

# Путь к скачанному архиву погоды
weather_path = Path(
    r"C:\Users\AsusVivo\Desktop\ФИНАЛЬНЫЙ ПРОЕКТ\04_Python\weather.csv"
)

# Показываем первые строки файла, чтобы определить его структуру
with weather_path.open(encoding='cp1251') as file:
    for _ in range(10):
        print(file.readline().rstrip())
```

```python
# Читаем только первые два столбца: дату и температуру
weather = pd.read_csv(
    weather_path,
    sep=';',
    encoding='cp1251',
    comment='#',
    usecols=[0, 1],
    index_col=False
)

weather.columns = ['Дата', 'T']

weather['Дата'] = pd.to_datetime(
    weather['Дата'],
    format='%d.%m.%Y %H:%M'
).dt.normalize()

weather['T'] = pd.to_numeric(weather['T'], errors='coerce')

# Средняя температура за каждый день
daily_weather = (
    weather.groupby('Дата')['T']
    .mean()
    .to_frame()
    .sort_index()
)

display(daily_weather.head())
```

```python
sales_weather = grouped_df.join(
    daily_weather,
    how='left',
    validate='one_to_one'
)

print('Дней без температуры:', sales_weather['T'].isna().sum())
display(sales_weather.head())
```

```python
ax = sales_weather.plot(
    y=['Количество продаж', 'T'],
    secondary_y=['T'],
    figsize=(14, 5),
    color=['steelblue', 'darkorange']
)

ax.set_title('Ежедневные продажи и температура в Астане')
ax.set_xlabel('Дата')
ax.set_ylabel('Количество проданных единиц')
ax.right_ax.set_ylabel('Средняя температура, °C')
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
ax = sales_weather['T'].plot(
    figsize=(14, 4),
    color='darkorange'
)

ax.set_title('Средняя дневная температура в Астане')
ax.set_xlabel('Дата')
ax.set_ylabel('Температура, °C')
ax.axhline(0, color='gray', linestyle='--', linewidth=1)
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
